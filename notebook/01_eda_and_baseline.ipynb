# Walmart -- Prédiction des ventes hebdomadaires

Problématique : comment prédire les ventes hebdomadaires des magasins Walmart à partir de variables économiques et calendaires ?

Approche :
1. EDA et nettoyage des données (valeurs manquantes, outliers)
2. Feature engineering (extraction des composantes de date)
3. Pipeline de preprocessing (imputation, scaling, encoding)
4. Modélisation : Régression Linéaire, Ridge, Lasso
5. Comparaison des performances

Métrique principale : R2 et RMSE sur le jeu de test.

## 1. Chargement et audit des données

Importation des librairies et chargement du dataset brut.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import (
    r2_score, mean_absolute_error, mean_squared_error
)

# Configuration de l'affichage
pd.set_option("display.max_columns", None)
pd.options.display.float_format = "{:.2f}".format
warnings.filterwarnings("ignore")
```

```python
# Chargement du dataset brut
df = pd.read_csv("../data/raw/Walmart_Store_sales.csv")
print(
    f"Dataset : {df.shape[0]} lignes, "
    f"{df.shape[1]} colonnes"
)
display(df.head())
df.info()
```

Statistiques descriptives pour détecter les anomalies et comprendre les distributions.

```python
# Statistiques descriptives du dataset
df.describe()
```

Focus sur la variable cible `Weekly_Sales` : distribution et valeurs extrêmes.

```python
# Distribution de la variable cible
df["Weekly_Sales"].dropna().describe()
```

### 1.1 Nettoyage de la target (`Weekly_Sales`)

On ne fait jamais d'imputation sur la target : suppression des lignes manquantes.

```python
# Suppression des lignes ou la target est manquante
missing_target = df["Weekly_Sales"].isnull().sum()
print(f"Target manquante : {missing_target} lignes")

df = df.dropna(subset=["Weekly_Sales"])
print(f"Apres suppression : {df.shape}")
```

### 1.2 Feature engineering : dates

Extraction de Year, Month, Day, DayOfWeek depuis la colonne `Date`, puis suppression de la colonne originale.

```python
# Conversion en datetime et extraction des composantes
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.dayofweek

df = df.drop("Date", axis=1)
display(df.head())
```

### 1.3 Suppression des outliers (règle 3 sigmas)

Exclusion des valeurs hors [moyenne - 3*écart-type, moyenne + 3*écart-type] pour Temperature, Fuel_Price, CPI et Unemployment.

```python
# Suppression des outliers par la regle des 3 sigmas
numeric_features = [
    "Temperature", "Fuel_Price", "CPI", "Unemployment"
]

print(f"Avant outliers : {df.shape}")

for col in numeric_features:
    mean = df[col].mean()
    std = df[col].std()
    lower_bound = mean - 3 * std
    upper_bound = mean + 3 * std

    mask_outliers = (
        (df[col] >= lower_bound) & (df[col] <= upper_bound)
    )
    df = df.loc[mask_outliers | df[col].isnull(), :]

    print(
        f"  {col} : [{lower_bound:.2f}, {upper_bound:.2f}]"
    )

print(f"Apres outliers : {df.shape}")
```

### 1.4 Distribution de la variable cible

Histogramme et boxplot de `Weekly_Sales` après nettoyage.

```python
# Distribution de la variable cible
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(
    df["Weekly_Sales"], bins=25,
    color="steelblue", edgecolor="white"
)
axes[0].set_title("Distribution de Weekly_Sales")
axes[0].set_xlabel("Weekly Sales ($)")

sns.boxplot(
    x=df["Weekly_Sales"], ax=axes[1], color="steelblue"
)
axes[1].set_title("Boxplot Weekly_Sales")

plt.tight_layout()
plt.savefig(
    "../assets/images/distribution_target.png",
    dpi=150, bbox_inches="tight"
)
plt.show()
```

## 3. Preprocessing Scikit-Learn

### 3.1 Séparation X/Y et train/test split (80/20)

Séparation des features et de la target, puis découpe en jeux d'entraînement et de test.

```python
# Separation target / features
Y = df["Weekly_Sales"]
X = df.drop("Weekly_Sales", axis=1)

# Split train/test 80/20
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

print(f"X_train : {X_train.shape}, X_test : {X_test.shape}")
```

### 3.2 Pipeline de preprocessing (ColumnTransformer)

- Numériques (Temperature, Fuel_Price, CPI, Unemployment, Year, Month, Day, DayOfWeek) : imputation par la moyenne + StandardScaler.
- Catégorielles (Store, Holiday_Flag) : imputation par le mode + OneHotEncoder (`drop='first'` pour éviter la colinéarité).

```python
# Definition des colonnes par type
numeric_features = [
    "Temperature", "Fuel_Price", "CPI",
    "Unemployment", "Year", "Month", "Day", "DayOfWeek"
]
categorical_features = ["Store", "Holiday_Flag"]

# Pipeline pour les variables numeriques
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()),
])

# Pipeline pour les variables categorielles
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(drop="first")),
])

# Assemblage dans un ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer,
         categorical_features),
    ]
)
```

### 3.3 Application du preprocessing

Le `fit` se fait uniquement sur le train pour éviter le data leakage.

```python
# fit sur train uniquement (eviter le data leakage)
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)

print(
    f"X_train apres preprocessing : {X_train.shape}"
)
```

## 4. Modèle Baseline : Régression Linéaire

Entraînement d'un modèle de régression linéaire simple comme point de référence.

```python
# Entrainement du modele baseline
regressor = LinearRegression()
regressor.fit(X_train, Y_train)
```

### 4.1 Évaluation des performances

Calcul du R2, MAE et RMSE sur les jeux d'entraînement et de test.

```python
# Predictions et metriques
Y_train_pred = regressor.predict(X_train)
Y_test_pred = regressor.predict(X_test)

r2_train = r2_score(Y_train, Y_train_pred)
r2_test = r2_score(Y_test, Y_test_pred)
mae_test = mean_absolute_error(Y_test, Y_test_pred)
rmse_test = np.sqrt(
    mean_squared_error(Y_test, Y_test_pred)
)

print("=== Regression Lineaire ===")
print(f"R2  Train : {r2_train:.3f} | Test : {r2_test:.3f}")
print(f"MAE Test  : {mae_test:,.0f} $")
print(f"RMSE Test : {rmse_test:,.0f} $")
```

Le R2 train (0.977) est nettement supérieur au R2 test (0.891), ce qui indique un léger overfitting. La régression linéaire capture bien la structure générale des données, mais l'écart train/test suggère que la régularisation pourrait améliorer la généralisation.

### 4.2 Réel vs Prédit (test set)

Visualisation des prédictions vs les valeurs réelles pour évaluer la qualité du modèle.

```python
# Graphique reel vs predit
plt.figure(figsize=(8, 6))
sns.scatterplot(x=Y_test, y=Y_test_pred, alpha=0.6)
plt.plot(
    [Y_test.min(), Y_test.max()],
    [Y_test.min(), Y_test.max()],
    "r--"
)
plt.xlabel("Valeurs reelles")
plt.ylabel("Valeurs predites")
plt.title(
    "Reel vs Predit - Regression Lineaire (test set)"
)
plt.tight_layout()
plt.savefig(
    "../assets/images/real_vs_predicted.png",
    dpi=150, bbox_inches="tight"
)
plt.show()
```

### 4.3 Interprétation des coefficients

Identification des features les plus influentes dans le modèle linéaire.

```python
# Extraction et tri des coefficients
feature_names_out = preprocessor.get_feature_names_out()

coefs = pd.DataFrame(
    index=feature_names_out,
    data=regressor.coef_,
    columns=["Coefficient"],
)
coefs["Abs"] = coefs["Coefficient"].abs()
coefs_sorted = coefs.sort_values("Abs", ascending=False)

print("Top 10 features :")
display(coefs_sorted.head(10))

# Visualisation des 10 features les plus importantes
plt.figure(figsize=(10, 6))
sns.barplot(
    y=coefs_sorted.head(10).index,
    x="Coefficient",
    data=coefs_sorted.head(10),
    palette="magma",
)
plt.title("Top 10 features par coefficient")
plt.tight_layout()
plt.savefig(
    "../assets/images/feature_importance.png",
    dpi=150, bbox_inches="tight"
)
plt.show()
```

## 5. Régularisation

### 5.1 Ridge Regression

Ajout d'une pénalité L2 (alpha * somme des beta^2) pour réduire l'overfitting observé (écart train/test).

```python
# Test avec alpha=100 (volontairement eleve pour voir l'effet)
ridge_100 = Ridge(alpha=100)
ridge_100.fit(X_train, Y_train)

Y_test_pred_ridge100 = ridge_100.predict(X_test)

mae_ridge100 = mean_absolute_error(
    Y_test, Y_test_pred_ridge100
)
rmse_ridge100 = np.sqrt(
    mean_squared_error(Y_test, Y_test_pred_ridge100)
)

print("=== Ridge (alpha=100) ===")
print(
    f"R2  Train : {ridge_100.score(X_train, Y_train):.3f}"
    f" | Test : {ridge_100.score(X_test, Y_test):.3f}"
)
print(f"MAE Test  : {mae_ridge100:,.0f} $")
print(f"RMSE Test : {rmse_ridge100:,.0f} $")
print(
    "-> Alpha trop eleve : le modele est trop penalise, "
    "d'ou l'effondrement du R2."
)
```

### 5.2 Optimisation de alpha par GridSearchCV (Ridge)

Recherche du meilleur alpha parmi une grille de valeurs candidates.

```python
# GridSearchCV pour trouver le meilleur alpha
params = {
    "alpha": [0.01, 0.1, 1, 10, 50, 100, 500, 1000]
}

grid_ridge = GridSearchCV(
    Ridge(), param_grid=params, cv=5, scoring="r2"
)
grid_ridge.fit(X_train, Y_train)

best_ridge = grid_ridge.best_estimator_
Y_test_pred_best_ridge = best_ridge.predict(X_test)

best_alpha = grid_ridge.best_params_["alpha"]
mae_best_ridge = mean_absolute_error(
    Y_test, Y_test_pred_best_ridge
)
rmse_best_ridge = np.sqrt(
    mean_squared_error(Y_test, Y_test_pred_best_ridge)
)

print(f"=== Ridge optimise (alpha={best_alpha}) ===")
print(f"R2 CV moyen : {grid_ridge.best_score_:.3f}")
print(
    f"R2  Train : "
    f"{best_ridge.score(X_train, Y_train):.3f}"
    f" | Test : {best_ridge.score(X_test, Y_test):.3f}"
)
print(f"MAE Test  : {mae_best_ridge:,.0f} $")
print(f"RMSE Test : {rmse_best_ridge:,.0f} $")
```

### 5.3 Lasso Regression (pénalité L1)

Contrairement à Ridge (L2), Lasso peut mettre des coefficients exactement à 0, ce qui assure une sélection de features automatique.

```python
# GridSearchCV pour Lasso
params_lasso = {
    "alpha": [0.01, 0.1, 1, 10, 50, 100, 500, 1000]
}

grid_lasso = GridSearchCV(
    Lasso(max_iter=10000),
    param_grid=params_lasso,
    cv=5,
    scoring="r2",
)
grid_lasso.fit(X_train, Y_train)

best_lasso = grid_lasso.best_estimator_
Y_test_pred_lasso = best_lasso.predict(X_test)

best_alpha_lasso = grid_lasso.best_params_["alpha"]
mae_lasso = mean_absolute_error(
    Y_test, Y_test_pred_lasso
)
rmse_lasso = np.sqrt(
    mean_squared_error(Y_test, Y_test_pred_lasso)
)

print(
    f"=== Lasso optimise "
    f"(alpha={best_alpha_lasso}) ==="
)
print(f"R2 CV moyen : {grid_lasso.best_score_:.3f}")
print(
    f"R2  Train : "
    f"{best_lasso.score(X_train, Y_train):.3f}"
    f" | Test : {best_lasso.score(X_test, Y_test):.3f}"
)
print(f"MAE Test  : {mae_lasso:,.0f} $")
print(f"RMSE Test : {rmse_lasso:,.0f} $")

# Coefficients mis a 0 (features eliminees par Lasso)
n_zeros = (best_lasso.coef_ == 0).sum()
total_coefs = len(best_lasso.coef_)
print(
    f"\nFeatures eliminees par Lasso : "
    f"{n_zeros}/{total_coefs}"
)
```

## 6. Comparaison des modèles

Tableau récapitulatif des performances de tous les modèles entraînés.

```python
# Tableau comparatif des modeles
best_alpha_ridge = grid_ridge.best_params_["alpha"]
best_alpha_lasso = grid_lasso.best_params_["alpha"]

results = pd.DataFrame({
    "Modele": [
        "Regression Lineaire",
        f"Ridge (alpha={best_alpha_ridge})",
        f"Lasso (alpha={best_alpha_lasso})",
    ],
    "R2 Train": [
        r2_score(Y_train, regressor.predict(X_train)),
        best_ridge.score(X_train, Y_train),
        best_lasso.score(X_train, Y_train),
    ],
    "R2 Test": [
        r2_score(Y_test, Y_test_pred),
        best_ridge.score(X_test, Y_test),
        best_lasso.score(X_test, Y_test),
    ],
    "MAE Test": [
        mean_absolute_error(Y_test, Y_test_pred),
        mae_best_ridge,
        mae_lasso,
    ],
    "RMSE Test": [
        rmse_test,
        rmse_best_ridge,
        rmse_lasso,
    ],
})

results["R2 Train"] = results["R2 Train"].map(
    "{:.3f}".format
)
results["R2 Test"] = results["R2 Test"].map(
    "{:.3f}".format
)
results["MAE Test"] = results["MAE Test"].map(
    "${:,.0f}".format
)
results["RMSE Test"] = results["RMSE Test"].map(
    "${:,.0f}".format
)

display(results)
```

## 7. Conclusion

Meilleur modèle : Lasso (alpha=500) avec R2 test = 0.897, MAE = 151 358 $, RMSE = 188 738 $.

Observations :
- L'écart train/test modéré (0.977 vs 0.891 en linéaire) indique un léger overfitting, corrigé partiellement par la régularisation.
- Ridge (best alpha=0.01) apporte un gain marginal (R2 test 0.892).
- Lasso (best alpha=500) donne le meilleur R2 test (0.897) et élimine 1 feature sur 27, ce qui simplifie le modèle.
- Le Store est de loin la feature la plus prédictive (effet taille/localisation du magasin).

Limites :
- Dataset très petit (131 lignes après nettoyage) : résultats sensibles au split.
- OneHotEncoding de Store crée beaucoup de colonnes pour peu d'observations.
